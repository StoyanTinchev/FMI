```python
import math
import numpy as np

# Define the third-degree Legendre polynomial
def legendre_p3(x):
    return (1/2) * (5*x**3 - 3*x)

# The roots of the third-degree Legendre polynomial (symmetric about the origin)
nodes = np.array([-math.sqrt(3/5), 0, math.sqrt(3/5)])

# Weights for three-node Gaussian quadrature
weights = np.array([5/9, 8/9, 5/9])

# Map the nodes from the interval [-1, 1] to [0, pi/4] using the change of variables
a, b = 0, math.pi/4
mapped_nodes = (b-a)/2 * nodes + (b+a)/2

def integrand(x):
    return x**2 + np.cos(x)

# Evaluate the integrand at the mapped nodes
integrand_values = integrand(mapped_nodes)

# Perform the Gaussian quadrature
integral_approximation = (b-a)/2 * np.sum(weights * integrand_values)

integral_approximation
```

```python
import math
import numpy as np

def integrand(x):
    return np.sin(x)**2

# Trapezoidal rule function
def trapezoidal_rule(f, a, b, n):
    h = (b - a) / n
    integral = 0.5 * (f(a) + f(b))
    integral += np.sum(f(a + h * np.arange(1, n)))
    integral *= h
    return integral

# Initialize variables
a = 1  # lower limit of integration
b = 3  # upper limit of integration
epsilon = 1e-5  # desired precision
n = 1  # start with one trapezoid
integral_previous = 0
integral_current = trapezoidal_rule(integrand, a, b, n)

# Iterate until the difference between successive approximations is less than epsilon
while abs(integral_current - integral_previous) > epsilon:
    integral_previous = integral_current
    n *= 2  # double the number of trapezoids
    integral_current = trapezoidal_rule(integrand, a, b, n)

# The result is the current integral approximation
integral_current, n  # Return the approximation and the number of trapezoids used
```
