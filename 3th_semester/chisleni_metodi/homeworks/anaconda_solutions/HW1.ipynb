```python
import math
import numpy as np
import matplotlib.pyplot as plt

# Data
x_data = np.array([0.1, 0.2, 0.4, 0.6, 0.9, 1.3, 1.5, 1.7, 1.8])
y_data = np.array([0.75, 1.25, 1.45, 1.25, 0.85, 0.55, 0.35, 0.28, 0.18])

# function y = a * exp(b * x)
def model_func(x, a, b):
    return a * np.exp(b * x)

def sum_of_squares(residuals):
    return np.sum(residuals**2)

# manually search for a combination of a and b that minimizes the sum of squares of the residuals.
# Initialize variables to store the best parameters and the smallest sum of squares found
best_a = None
best_b = None
smallest_sum_of_squares = float('inf')

for a in np.linspace(0, 10, 200):
    for b in np.linspace(-10, 0, 200):
        # Calculate the residuals for this combination of a and b
        residuals = y_data - model_func(x_data, a, b)
        # Calculate the sum of squares
        ss = sum_of_squares(residuals)
        
        # If this sum of squares is smaller than any we've found before, store these parameters as the best
        if ss < smallest_sum_of_squares:
            best_a = a
            best_b = b
            smallest_sum_of_squares = ss
            
            
plt.scatter(x_data, y_data, color='red', label='Data Points')

x_model = np.linspace(min(x_data), max(x_data), 1000)
y_model = model_func(x_model, best_a, best_b)
plt.plot(x_model, y_model, label='Best Fit: y = {:.2f}e^({:.2f}x)'.format(best_a, best_b))

plt.title('Least Squares Fit to Data')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()

plt.show()

# Print the best parameters and the smallest sum of squares
best_a, best_b, smallest_sum_of_squares
```

```python
import math
import numpy as np
import matplotlib.pyplot as plt

# Transcribing the new data points from the image provided
x_data = np.array([1, 2, 3, 4, 5])
y_data = np.array([2.2, 2.8, 3.6, 4.5, 5.5])

# Define the model function y = a + b*x + c/x
def model_func(x, a, b, c):
    return a + b*x + c/x

def sum_of_squares(residuals):
    return np.sum(residuals**2)

# Initialize variables to store the best parameters and the smallest sum of squares
best_a = best_b = best_c = None
smallest_sum_of_squares = float('inf')

for a in np.linspace(0, 10, 100):  # Coarse search for a
    for b in np.linspace(0, 10, 100):  # Coarse search for b
        for c in np.linspace(0, 10, 100):  # Coarse search for c
            # Calculate the residuals for this combination of a, b, and c
            residuals = y_data - model_func(x_data, a, b, c)
            # Calculate the sum of squares
            ss = sum_of_squares(residuals)
            
            # If this sum of squares is smaller than any found before, store these parameters as the best
            if ss < smallest_sum_of_squares:
                best_a = a
                best_b = b
                best_c = c
                smallest_sum_of_squares = ss

                
plt.scatter(x_data, y_data, color='blue', label='Data Points')

x_model = np.linspace(min(x_data), max(x_data), 1000)
y_model = model_func(x_model, best_a, best_b, best_c)
plt.plot(x_model, y_model, label='Best Fit: y = {:.2f} + {:.2f}x + {:.2f}/x'.format(best_a, best_b, best_c))

plt.title('Least Squares Fit to Data')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()

plt.show()

best_a, best_b, best_c, smallest_sum_of_squares
```
